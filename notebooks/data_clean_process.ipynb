```python
import os
import pandas as pd
import rasterio

# 1. Load your master station directory containing the true coordinate profiles
stations_profile_csv = "/home/prasad/Ganga-River-stem_ES-Project/docs/stations.csv"

if not os.path.exists(stations_profile_csv):
    raise FileNotFoundError(f"Missing master coordinate reference file at: {stations_profile_csv}")

df_stations = pd.read_csv(stations_profile_csv)

# Isolate unique location values exactly matching your layout keys
df_geo = df_stations[['Station_name', 'Latitude', 'longitude']].drop_duplicates().reset_index(drop=True)

# 2. Setup your raw population raster folder and years loop
pop_dir = "/home/prasad/Ganga-River-stem_ES-Project/data/raw/worldpop"
years = [2000, 2005, 2010, 2015, 2020]

extracted_records = []
print("Extracting local 1km population density counts for station coordinates...")

# Iterate through each unique station location
for idx, row in df_geo.iterrows():
    name = row['Station_name']
    lon = row['longitude']
    lat = row['Latitude']
    
    # Extract values across every census milestone grid year
    for year in years:
        raster_name = f"ind_pd_{year}_1km.tif"
        raster_path = os.path.join(pop_dir, raster_name)
        
        if os.path.exists(raster_path):
            try:
                with rasterio.open(raster_path) as src:
                    # rasterio samples point coordinates as (longitude, latitude) index mapping vectors
                    coords = [(lon, lat)]
                    sample = list(src.sample(coords))
                    
                    # FIXED: Extracting the actual value from the internal NumPy array format safely
                    pop_value = float(sample[0][0])
                    
                    # Handle any null spatial border pixel arrays gracefully
                    if pop_value < 0 or pd.isna(pop_value):
                        pop_value = 0.0
                        
                    extracted_records.append({
                        "station_name": name,
                        "latitude": lat,
                        "longitude": lon,
                        "year": year,
                        "population_density_1km": round(pop_value, 2)
                    })
            except Exception as e:
                print(f"Error reading grid asset {raster_name} for station {name}: {e}")
        else:
            print(f"Warning: Missing data grid file asset: {raster_path}")

# 3. Compile output dataframe matrix and save to csv spreadsheet file
df_pop_final = pd.DataFrame(extracted_records)
output_path = "/home/prasad/Ganga-River-stem_ES-Project/data/processed/station_population_trends.csv"
df_pop_final.to_csv(output_path, index=False)

print(f"\nSuccess! Demographic data extraction complete. Saved file containing {len(df_pop_final)} records.")
print(f"Location: {output_path}")
display(df_pop_final.head(10))
```

```python
import geopandas as gpd
import pandas as pd
import os

# 1. Load your spatial assets and match the exact project path directories
rivers_shp = "/home/prasad/Ganga-River-stem_ES-Project/data/processed/ganga_rivers_clipped.shp"
stations_csv = "/home/prasad/Ganga-River-stem_ES-Project/docs/stations.csv"

rivers = gpd.read_file(rivers_shp).to_crs(7755)
st = pd.read_csv(stations_csv)

# 2. Convert your station points into a metric GeoDataFrame matching your case-sensitive headers
pts = gpd.GeoDataFrame(
    st, 
    geometry=gpd.points_from_xy(st['longitude'], st['Latitude']),
    crs=4326
).to_crs(7755)

# Verify that the discharge column exists inside your shapefile layout attributes
if "DIS_AV_CMS" not in rivers.columns:
    raise KeyError("DIS_AV_CMS column not found in HydroRIVERS shapefile attributes!")

rows = []

# 3. Spatial matching loop
for _, s in pts.iterrows():
    # Calculate distances to find all river segments within a 5 km buffer zone radius
    distances = rivers.distance(s.geometry)
    near = rivers[distances < 5000]
    
    if not near.empty:
        # Isolate the segment with the largest average flow (this isolates the main stem of the Ganga)
        main_segment = near.loc[near["DIS_AV_CMS"].idxmax()]
        
        rows.append({
            "station_name": s["Station_name"],
            "town": s["Town"],
            "flow_avg_cms": float(main_segment["DIS_AV_CMS"])
        })
    else:
        print(f"Warning: Station {s['Station_name']} did not snap to any river line within 5km.")

# 4. Save and verify trend output downstream
df_flow = pd.DataFrame(rows)
output_flow_csv = "/home/prasad/Ganga-River-stem_ES-Project/data/processed/station_avg_flow.csv"
df_flow.to_csv(output_flow_csv, index=False)

print(f"\nSuccess! Level 2 Complete. Average flow values saved to: {output_flow_csv}\n")
print("--- Extracted Natural Flow Summary (Verify Downstream Accumulation) ---")
print(df_flow.to_string(index=False))
```

```python
from pathlib import Path
import pandas as pd

# 1. Establish project directory configuration
ROOT = Path("/home/prasad/Ganga-River-stem_ES-Project")
wq_file = ROOT / "data/processed/wq_clean_final.csv"
ev_file = ROOT / "data/raw/events/events.csv"

wq = pd.read_csv(wq_file)
ev = pd.read_csv(ev_file)

# Ensure date columns are formatted as datetime objects for temporal comparisons
# Note: If your water quality file uses a different column name (like 'sampling_date'), swap it here
date_col = 'date' if 'date' in wq.columns else wq.columns[2] # Fallback to 3rd column if named differently
wq[date_col] = pd.to_datetime(wq[date_col], errors="coerce")

ev["start_date"] = pd.to_datetime(ev["start_date"])
ev["end_date"] = pd.to_datetime(ev["end_date"])

# Map your town locations directly to the long-form station names
town_to_station_map = {
    "haridwar": "ganga at har-ki-pauri ghat",
    "kanpur": "ganga at jajmau bridge",
    "prayagraj": "ganga at sangam",
    "allahabad": "ganga at sangam",
    "varanasi": "ganga at malviya bridge",
    "patna": "ganga at gandhi ghat"
}

# 2. Window validation function with variable padding constraints
def calculate_window_density(row, pad_days):
    town = str(row["place"]).lower().strip()
    target_station = town_to_station_map.get(town, "")
    
    if not target_station:
        return 0
        
    # Isolate records for this specific station
    s = wq[wq["station_name"].str.lower().str.strip() == target_station]
    
    # Calculate window parameters using timedeltas
    start_buffer = row["start_date"] - pd.Timedelta(days=pad_days)
    end_buffer = row["end_date"] + pd.Timedelta(days=pad_days)
    
    # Slice the dataframe strictly inside the dates
    in_window = s[(s[date_col] >= start_buffer) & (s[date_col] <= end_buffer)]
    return len(in_window)

# 3. Calculate metrics for both a tight 15-day window and a wider 30-day window
ev["n_samples_pad15"] = ev.apply(lambda r: calculate_window_density(r, 15), axis=1)
ev["n_samples_pad30"] = ev.apply(lambda r: calculate_window_density(r, 30), axis=1)

print("\n--- True Event-Window Sample Validation Table ---")
summary_cols = ["event_name", "place", "start_date", "n_samples_pad15", "n_samples_pad30"]
print(ev[summary_cols].to_string(index=False))
```

```python
import os
import pandas as pd
import numpy as np
from pathlib import Path
from scipy.stats import mannwhitneyu, spearmanr

# 1. Establish project absolute paths workspace environment
ROOT = Path("/home/prasad/Ganga-River-stem_ES-Project")
wq_file = ROOT / "data/processed/wq_clean_final.csv"
pop_file = ROOT / "data/processed/station_population_trends.csv"

# Load clean files matrix
wq = pd.read_csv(wq_file)
pop = pd.read_csv(pop_file)

print("--- 1. Diagnostic Checks (Season Footprints) ---")
print(wq["season"].value_counts())
print("\nRows per station-year footprint summary descriptor:")
print(wq.groupby(["station_name", "year"]).size().describe())

# 2. Standardize target column parameter formats to numeric values
cols = ["bod_max", "do_min", "fecal_col_max", "total_col_max"]
for c in cols:
    wq[c] = pd.to_numeric(wq[c], errors="coerce")

# Compress seasonal variations into annual median summaries for testing
wq_year = wq.groupby(["station_name", "year"])[cols].median().reset_index()

# Define historically verified Mega Event tracking years present in your data matrix span
mega = {
    "Ganga at Har-Ki-Pauri Ghat":,  # Dropped 2025 since data isn't out yet
    "Ganga at Sangam": [2013, 2019]                   # Dropped 2025
}

print("\n--- 2. Mann-Whitney U Statistical Testing (Event vs Non-Event Years) ---")
for station, years in mega.items():
    d = wq_year[wq_year["station_name"] == station]
    ev_d = d[d["year"].isin(years)]
    non = d[~d["year"].isin(years)]
    
    for c in cols:
        a = ev_d[c].dropna()
        b = non[c].dropna()
        
        if len(a) >= 2 and len(b) >= 2:
            u, p = mannwhitneyu(a, b, alternative="two-sided")
            print(f"{station[:25]}... | {c:13}: Event Median={a.median():6.1f} (n={len(a)}), "
                  f"Normal Median={b.median():6.1f} (n={len(b)}), p-val={p:.3f}")
        else:
            print(f"{station[:25]}... | {c:13}: Too few points to run test (n={len(a)} vs n={len(b)})")

print("\n--- 3. Merging Clean Water Quality Metrics with Population Trends ---")
# Ensure data validation tracking strings link up seamlessly without trailing spaces
pop["station_name"] = pop["station_name"].str.strip()
wq_year["station_name"] = wq_year["station_name"].str.strip()

# Compile the final data matrix cross-join tracking sheet
df_final_analysis = pd.merge(wq_year, pop, on=["station_name", "year"], how="inner")
analysis_output_path = ROOT / "data/processed/wq_demographic_merged.csv"
df_final_analysis.to_csv(analysis_output_path, index=False)
print(f"Success! Merged dataset safely compiled with {len(df_final_analysis)} active trend rows.")
print(f"Saved to location: {analysis_output_path}")

print("\n--- 4. Spearman Rank Correlation (Water Quality vs Population Density) ---")
for c in ["bod_max", "do_min", "fecal_col_max"]:
    sub = df_final_analysis[[c, "population_density_1km"]].dropna()
    if len(sub) >= 3:
        coef, p_val = spearmanr(sub[c], sub["population_density_1km"])
        print(f"Spearman Link -> {c:13} vs Population Density: Coef={coef:.3f}, p-val={p_val:.3f}")
```
